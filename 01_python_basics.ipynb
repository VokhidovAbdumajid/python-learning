```python
pulse = [120, 85, 200, 150, 95, 110, 80]

total = sum(pulse)
count = len(pulse)

average = total / count

print("Average pulse:", average)
```

```python
pulse = [120, 85, 200, 150, 95, 110, 80]

print("Count:", len(pulse))
print("Maximum:", max(pulse))
print("Minimum:", min(pulse))
print("Average:", sum(pulse) / len(pulse))
```
